# GPU connection check

Run this notebook with the Python kernel that has PyTorch installed. On an HPC cluster, start Jupyter from an interactive GPU allocation (or use the cluster's documented GPU notebook service); a notebook running on a login or CPU node will usually report that CUDA is unavailable.

This checks that PyTorch can see a CUDA GPU and complete a small matrix multiplication.

```python
import platform
import sys

print(f"Python: {platform.python_version()} ({sys.executable})")

try:
    import torch
except ImportError as exc:
    raise ImportError(
        "PyTorch is not installed in this notebook's Python environment. "
        "Select the cluster's PyTorch kernel/module or install a build compatible with its CUDA setup."
    ) from exc

print(f"PyTorch: {torch.__version__}")
print(f"PyTorch CUDA build: {torch.version.cuda}")
print(f"CUDA available: {torch.cuda.is_available()}")
```

```python
if not torch.cuda.is_available():
    print("No CUDA GPU is visible to this notebook process. Check that this kernel is running inside a GPU allocation.")
else:
    device = torch.device("cuda:0")
    properties = torch.cuda.get_device_properties(device)
    print(f"GPU: {properties.name}")
    print(f"GPU memory: {properties.total_memory / 1024**3:.2f} GiB")

    # A small operation confirms that work can execute on the GPU.
    a = torch.rand((2000, 2000), device=device)
    b = a @ a
    torch.cuda.synchronize(device)
    print(f"GPU computation: OK; result device={b.device}, shape={tuple(b.shape)}")
```
